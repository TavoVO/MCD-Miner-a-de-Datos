```python
estudiantes = [
    [9.5, 10000, 0, 'SI'],
    [9.0, 12000, 1, 'SI'],
    [8.5, 14000, 0, 'SI'],
    [8.8, 13000, 2, 'NO'],
    [7.0, 10000, 0, 'NO'],
    [6.5, 8000, 0, 'NO'],
    [8.2, 20000, 0, 'NO'],
    [9.1, 25000, 1, 'NO']
    ]
```

```python
def mostrar_arbol():
    print("Arbol Actual:")

    print("¿Promedio >= 8 ?")
    print("|")
    print("|-- NO -> Beca = NO")
    print("|")
    print("|-- Sí")
    print("     |")
    print("     |-- ¿Ingresos < 15000 ?")
    print("     |")
    print("     |-- NO -> Beca = NO")
    print("     |")
    print("     |-- Sí")
    print("     |    |")
    print("     |    |-- ¿Número de Reprobadas <= 1 ?")
    print("     |    |            |")
    print("     |    |            |-- Sí -> Beca = SI")
    print("     |    |            |")
    print("     |    |            |-- NO -> Beca = NO")
```

```python
print("Paso 1")
print("División por Promedio")
mostrar_arbol()
```

```python
def otorgar_beca(promedio, ingresos, reprobadas):
    # nodo raíz
    if promedio < 8:
        return 'NO'
    # segundo nivel
    if ingresos >= 15000:
        return 'NO'
    # tercer nivel
    if reprobadas > 1:
        return 'NO'
    
    return 'NO'
```

```python
for estudiante in estudiantes:
    promedio = estudiante[0]
    ingresos = estudiante[1]
    reprobadas = estudiante[2]

    resultado = otorgar_beca(promedio, ingresos, reprobadas)

    print(f"Estudiante con promedio {promedio}, ingresos {ingresos}, reprobadas {reprobadas} -> Beca: {resultado}")
    print("Predicción",resultado)
    print("--------------------------------------------------")
```
