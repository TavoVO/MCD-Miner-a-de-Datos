<h3>Regresión Lineal<h3>

Con python

```python
x = [1 ,2 ,3 ,4 ,5]
y = [52 ,55 ,61 ,66 ,72]

n = len ( x )
# Calculo de las medias
media_x = sum ( x ) / n
media_y = sum ( y ) / n

numerador = 0
denominador = 0

# Calculo del numerador y denominador para la pendiente
for i in range ( n ) :
    numerador += ( x [ i ] - media_x ) *( y [ i ] - media_y )
    denominador += ( x [ i ] - media_x ) **2

# Calculo de los parametros
beta1 = numerador / denominador
beta0 = media_y - beta1 * media_x
print ( " Pendiente : " , beta1 )
print ( " Intercepto : " , beta0 )

# Prediccion
x_pred = 6
y_pred = beta0 + beta1 * x_pred
print ( f" Prediccion para x = {x_pred} : y = {y_pred} " )
#Predecir 10 más y graficarlos
x_pred_values = np.arange(6, 16) # Valores de x para predecir
y_pred_values = beta0 + beta1 * x_pred_values # Valores de y predichos
# Graficar los puntos originales y las predicciones
plt.figure(figsize=(8,5))
plt.scatter(x, y, color='blue', label='Datos Originales')
plt.plot(x_pred_values, y_pred_values, color='red', marker='o', label='Predicciones')
plt.title('Predicción de valores adicionales')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```

<h3>Tarea 2 - Tratamiento de duplicados y nulos<h3>


Para los valores nulos, donde son valores númericos como calificación de producto y precio unitario, se puede reemplazar por la media o mediana de esa columna. Para los valores nulos en columnas categóricas como categoría de producto, se puede reemplazar por la moda o el valor más frecuente en esa columna. En algunos casos, también se podría considerar eliminar filas con valores nulos si representan una pequeña proporción del dataset y no son críticos para el análisis.

```python
# Rellenar valores nulos con la media de cada columna
# Poner en el precio unitario la media con respecto a la subcategoria de producto
df_ventas_limpio['precio_unitario'] = df_ventas_limpio.groupby('subcategoria')['precio_unitario'].transform(lambda x: x.fillna(x.mean()))

# Poner en la calificación del cliente la media de la columna con respecto al id_cliente
df_ventas_limpio['calificacion_cliente'] = df_ventas_limpio.groupby('id_cliente')['calificacion_cliente'].transform(lambda x: x.fillna(x.mean()))
```

```python
df_ventas_limpio.isnull().sum()
```

```python
# Verificar si hay valores nulos en la columna 'calificacion_cliente' después de la imputación
promedios = df_ventas_limpio.groupby('id_cliente')['calificacion_cliente'].mean()
df_revision = (
    df_ventas_limpio[df_ventas_limpio['calificacion_cliente'].isna()]
    .assign(promedio_cliente=lambda x: x['id_cliente'].map(promedios))
)
df_revision
```

```python
# Asignar el promedio de calificación del cliente a los registros que aún tienen valores nulos
df_ventas_limpio.loc[df_ventas_limpio['calificacion_cliente'].isna(), 'calificacion_cliente'] = df_ventas_limpio['calificacion_cliente'].mean()
```

<h3>Tarea 4 - Detección y tratamiento de outliers<h3>

```python
#Generar un boxplot y un Histograma dataset original vs limpio para la columna precio_unitario
# Usar el metodo del z-score para identificar outliers en la columna precio_unitario y también el IQR
from scipy import stats
import matplotlib.pyplot as plt
z_scores = stats.zscore(df_ventas_limpio['precio_unitario'])
outliers = df_ventas_limpio[np.abs(z_scores) > 3]
Q1 = df_ventas_limpio['precio_unitario'].quantile(0.25)
Q3 = df_ventas_limpio['precio_unitario'].quantile(0.75)
IQR = Q3 - Q1
outliers_iqr = df_ventas_limpio[(df_ventas_limpio['precio_unitario'] < (Q1 - 1.5 * IQR)) | (df_ventas_limpio['precio_unitario'] > (Q3 + 1.5 * IQR))]
# Mostrar los outliers identificados
print("Outliers identificados con Z-score:")
print(outliers[['precio_unitario']])
print("Outliers identificados con IQR:")
print(outliers_iqr[['precio_unitario']])
# Boxplot
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.boxplot(df_ventas['precio_unitario'].dropna(), vert=False)
plt.title('Boxplot Precio Unitario - Original')
plt.subplot(1, 2, 2)
plt.boxplot(df_ventas_limpio['precio_unitario'], vert=False)
plt.title('Boxplot Precio Unitario - Limpio')
plt.tight_layout()
plt.show()
# Histograma
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.hist(df_ventas['precio_unitario'].dropna(), bins=20, color='blue', alpha=0.7)
plt.title('Histograma Precio Unitario - Original')
plt.subplot(1, 2, 2)
plt.hist(df_ventas_limpio['precio_unitario'], bins=20, color='green', alpha=0.7)
plt.title('Histograma Precio Unitario - Limpio')
plt.tight_layout()
plt.show()
```

```python
# Contar número de outliers identificados por cada método
num_outliers_z = outliers.shape[0]
num_outliers_iqr = outliers_iqr.shape[0]
print(f"Número de outliers identificados con Z-score: {num_outliers_z}")
print(f"Número de outliers identificados con IQR: {num_outliers_iqr}")
```

```python
# Eliminar los outliers identificados por el método del Z-score
df_ventas_sin_outliers_z = df_ventas_limpio[np.abs(z_scores) <= 3]
# Eliminar los outliers identificados por el método del IQR
df_ventas_sin_outliers_iqr = df_ventas_limpio[(df_ventas_limpio['precio_unitario'] >= (Q1 - 1.5 * IQR)) & (df_ventas_limpio['precio_unitario'] <= (Q3 + 1.5 * IQR))]
# Mostrar el número de registros después de eliminar los outliers
print(f"Número de registros después de eliminar outliers con Z-score: {df_ventas_sin_outliers_z.shape[0]}")
print(f"Número de registros después de eliminar outliers con IQR: {df_ventas_sin_outliers_iqr.shape[0]}")
# Obtener el dataset limpio final después de eliminar los outliers con el método del IQR
df_ventas_limpio_v2 = df_ventas_sin_outliers_iqr
```

```python
# Eliminar los valores de la columnna catidad donde son valores negativos
df_ventas_limpio_v3 = df_ventas_limpio_v2[df_ventas_limpio_v2['cantidad'] >= 0]
# Mostrar el número de registros después de eliminar los valores negativos en cantidad
print(f"Número de registros después de eliminar valores negativos en cantidad: {df_ventas_limpio_v3.shape[0]}")
```

<p>Porqué no se pueden imputar las fials con valores negativos?<p>

<p>Bueno, pues porque la imputación solo nos sirve para los casos en los que hay valores nulos o donde faltan datos, no para reemplazar valores y dada la falta de contexto pues es mejor eliminarlos para no valores que afecten<p>

```python
# Del total_venta checar los valores = 0 y ver los distintos casos para ver que se puede hacer
# Cantidad de registros donde total_venta es igual a 0
print("\nCantidad de registros donde total_venta es igual a 0:")
print(df_ventas_limpio_v3[df_ventas_limpio_v3['total_venta'] == 0].shape[0])

# Porcentaje de registros donde total_venta es igual a 0
print("\nPorcentaje de registros donde total_venta es igual a 0:")
print(df_ventas_limpio_v3[df_ventas_limpio_v3['total_venta'] == 0].shape[0] / df_ventas_limpio_v3.shape[0] * 100)

# Verificar los valores de precio_unitario para los registros donde total_venta es igual a 0
print("\nValores de precio_unitario para registros con total_venta = 0:")
print(df_ventas_limpio_v3[df_ventas_limpio_v3['total_venta'] == 0]['precio_unitario'].value_counts())

# Verificar los valores de cantidad para los registros donde total_venta es igual a 0
print("\nValores de cantidad para registros con total_venta = 0:")
print(df_ventas_limpio_v3[df_ventas_limpio_v3['total_venta'] == 0]['cantidad'].value_counts())
```

```python
# Si tiene precio unitario y cantidad, se hace la multiplicación y se actualiza el valor de total_venta
df_ventas_limpio_v3.loc[(df_ventas_limpio_v3['total_venta'] == 0) & (df_ventas_limpio_v3['precio_unitario'] > 0) & (df_ventas_limpio_v3['cantidad'] > 0), 'total_venta'] = df_ventas_limpio_v3['precio_unitario'] * df_ventas_limpio_v3['cantidad']
# Cantidad de registros donde total_venta es igual a 0 después de la actualización
print("\nCantidad de registros donde total_venta es igual a 0 después de la actualización:")
print(df_ventas_limpio_v3[df_ventas_limpio_v3['total_venta'] == 0].shape[0])
```

<p>Justificación:<p>
<p>Puesto que se tiene los datos de precio unitario y cantidad de la transacción se hace la multiplicación de precio por cantidad para sacar el total de la venta<p>
